# Naming Conventions

## A variable can have a short name (like x and y) or a more descriptive name (age, carname, total_volume).

### Rules for Python variables:

--> A variable name must start with a letter or the underscore character
--> A variable name cannot start with a number
--> A variable name can only contain alpha-numeric characters and underscores (A-z, 0-9, and _ )
--> Variable names are case-sensitive (age, Age and AGE are three different variables)
--> A variable name cannot be any of the Python keywords.

```python
# Valid Declaration of variables

first_name="Praveen"
last_name="Pakanati"
age=22
year=2004
_branch="cse"
my_num1=9154695089

print(first_name)
print(last_name)
print(age)
print(year)
print(_branch)
print(my_num1)


```

```python
# Invalid Declaration of variables

first-name="praveen"
last-name="pakanati"
1age=22
@year=2004

```

# understanding Variable Types
--> Python is dynamically typed ,means type of a variable is determined at runtime 


```python
age=25 #int
name="praveen" #str
grade=8.0 #float
is_fresher=True #boolean

#Type checking
print(type(name))
print(type(age))
print(type(grade))
print(type(is_fresher))
```

# Type conversion
--> You can convert from one type to another type using int(),str(),float().


```python
#Type conversion
age=22
print(type(age))

# Converts int to float

print(type(float(age)))
```

```python
name="praveen"
print(type(name))

#cannot convert str to int 

print(type(int(name)))
```

```python
age=25
print(type(age))

#int to bool

print(type(bool(age)))
```

# Dynamically Typed 

--> Python allows the type of variable to change as the progran executes.
--> you need not to mention the type of a variable unlike other languages.



```python
var=10 #int
print(var,type(var))

var="Hello"
print(var,type(var))

var=3.14
print(var,type(var))
```

# Input()

--> The input() function in Python is used to take input from the user. It waits for the user to type something on keyboard and once user presses Enter, it returns the value.

--> By default, input() always returns data as a string, even if you enter numbers. If you want other types like integer (int) or floating-point (float), you have to convert (typecast) the value explicitly.

```python
name=input("enter Your Name:")
print(type(name))
print(name)
```

```python
## Simple calculator

num1=float(input("Enter First Number:"))
num2=float(input("Enter second Number:"))

sum=num1+num2
sub=num1-num2
divide=num1/num2
multiply=num1*num2

print("Sum:",sum)
print("Subtraction:",sub)
print("Division:",divide)
print("Multiplication:",multiply)
```
