# List Slicing
List slicing allows you to create a new list by selecting a range of elements from an existing list.

## Syntax
list[start : stop : step]
start → Starting index (included)
stop → Ending index (excluded)
step → Number of positions to move each time (optional)

Remember: The stop index is not included.

```python
numbers = [10, 20, 30, 40, 50, 60]

print(numbers[1:4])
print(numbers[:3]) # start from beginning
print(numbers[2:]) # slice to the end
print(numbers[2:6])
print(numbers[:]) # copies entire list
print(numbers[-4:-1]) # negative indexing
print(numbers[::3])
print(numbers[::-1]) # reverse list
```
