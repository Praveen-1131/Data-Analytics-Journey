```python
# Looping through "for"
fruits = ["Apple", "Banana", "Mango"]

for fruit in fruits:
    print(fruit)
```

```python
# looping using range()
fruits = ["Apple", "Banana", "Mango"]

for i in range(len(fruits)):
    print(i, fruits[i])
```

```python
# using while loop
numbers = [10, 20, 30, 40]

i = 0

while i < len(numbers):
    print(numbers[i])
    i += 1
```

```python
## enumerate()-->returns both index and values
fruits = ["Apple", "Banana", "Mango"]

for index, fruit in enumerate(fruits):
    print(index, fruit) 
```

```python
## Looping through Strings
names = ["Praveen", "Rahul", "Amit"]

for name in names:
    print("Hello", name)
```

```python
# looping through nested lists
matrix = [
    [1, 2],
    [3, 4],
    [5, 6]
]

for row in matrix:
    for value in row:
        print(value)
```
