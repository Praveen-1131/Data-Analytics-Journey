# List Comprehension
--> List comprehension offers a shorter syntax when you want to create a new list based on the values of an existing list.

# Syntax
new_list = [expression for item in iterable]
** expression → What you want to add to the new list.
** item → Each element from the iterable.
** iterable → The object you're looping through.

```python
# without list comprehension
squares = []

for i in range(1, 6):
    squares.append(i * i)

print(squares)
```

```python
# using List Comprehension
squares = [i * i for i in range(1, 6)]

print(squares)
```

```python
# convert strings to upper case
names = ["praveen", "rahul", "amit"]

upper_names = [name.upper() for name in names]

print(upper_names)
```

# List Comprehension with if
syntax:
new_list = [expression for item in iterable if condition]

```python
# example using if 
evens = [x for x in range(1, 11) if x % 2 == 0]

print(evens)
```

```python
numbers = [10, 25, 18, 40, 5]

result = [num for num in numbers if num > 20]

print(result)
```

# List Comprehension with if-else
Syntax:
new_list = [expression_if_true if condition else expression_if_false for item in iterable]

```python
numbers = [1, 2, 3, 4, 5]

result = ["Even" if x % 2 == 0 else "Odd" for x in numbers]

print(result)
```

```python
numbers = [1, 2, 3, 4, 5]

result = [x * x for x in numbers if x % 2 == 0]

print(result)
```

# Nested List Comprehension

```python
matrix = [
    [1, 2],
    [3, 4],
    [5, 6]
]

result = [num for row in matrix for num in row]

print(result)
```
