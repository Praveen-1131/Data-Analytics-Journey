# List Methods:
Methods are used to modify the data in the list
List contains different kind of methods

# Adding Elements


```python
#1.Append()
## Append used to add element at the end of the list.
fruits = ["Apple", "Banana"]

fruits.append("Mango")

print(fruits)
```

```python
#2.Insert()
## Insert() is used to add the element at the particular index in the list 
names = ["Ram", "Ravi"]

names.insert(1, "Praveen")

print(names)
```

```python
# 3.Extend()
## Add the multiple elements at the end of the list 
numbers = [1, 2]

numbers.extend([3, 4, 5])

print(numbers)
```

# Removing Elements

```python
# 1.remove()
## removes the first occurence of the element from the list
numbers = [10, 20, 30, 20]

numbers.remove(20)

print(numbers)
```

```python
# 2.pop()
## pop() removes and return element from the particular index if no index found removes the last element by default
numbers = [10, 20, 30]

numbers.pop()
numbers.pop(1)

print(numbers)
```

```python
# 3.clear()
## clears everything from the list and gives empty list as output
numbers = [1, 2, 3]

numbers.clear()

print(numbers)
```

## Other List Methods

```python
# 1.Index()
## index() returns the index of the element from the list
numbers = [10, 20, 30]

print(numbers.index(20))
```

```python
# 2.count()
## counts the occurence of the element in the list
numbers = [1, 2, 2, 3, 2]

print(numbers.count(2))
```

```python
# 3.sort()
## sort used to sort the elements in ascending order by default
numbers = [5, 1, 4, 2]

numbers.sort()
print(numbers)
numbers.sort(reverse=True)

print(numbers)

print(numbers)
```

```python
# 4.reverse()
## reverse the elements in the list
numbers = [1, 2, 3]

numbers.reverse()

print(numbers)
```

```python
# 5.copy()
## creates the shallow copy of the list
list1 = [10, 20, 30]

list2 = list1.copy()

print(list2)
```
