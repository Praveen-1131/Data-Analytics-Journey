# List:
## 1. What is a List?

--> A list is an ordered, mutable (changeable) collection that can store multiple values of different data types.
Mutable: list elements can be changed, updated, added, or removed after the list is created.
Ordered: elements maintain the order in which they are inserted.
Index-based: elements are accessed using their position, starting from index 0.


```python
##Creating a List using []
a=[1,3,5,6,7,]
print(a)
```

```python
## Creating a list using list()
l=list((1,"Apple,3.16,True"))
print(l)

str=list("Hello")
print(str)
```

# 2.Accessing Elements
Elements in a list are accessed using indexing. Python uses zero-based indexing, meaning a[0] represents the first element. Negative indexing is also supported, where -1 accesses the last element.

```python
#using Positive Index
numbers=[1,4,7,9,4,12,33]
print(numbers)
print(numbers[0])
print(numbers[4])
print(numbers[1])
print(numbers[3])
```

```python
#using Negative Index
lst=[1,2,"banana",3.14,False,"hello",90]
print(lst[-1])
print(lst[-5])
print(lst[-2])
```

```python
##Modifying List Elements
fruits=["apple","banana","guvava","kiwi","orange"]
print(fruits)

fruits[1]="Watermelon"
print(fruits)
fruits[-1]="mango"
print(fruits)

fruits[4]="papaya"
print(fruits)
```
